```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_tavily import TavilySearch
from langchain.agents import create_agent

load_dotenv()

searchTool = TavilySearch(max_results=3)

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=os.getenv("GEMINI_API_KEY"),
    temperature=0.7,
    max_output_tokens=150
)

print("Setup complete!")
```

```python
agent = create_agent(
    model=llm,
    tools=[searchTool],
    system_prompt="You are a helpful assistant. Answer the user's question briefly and accurately using the available tools."
)

res = agent.invoke({
    "messages": [("human", "what is the population of bhopal?")]
})

print(res["messages"][-1].content)
```
